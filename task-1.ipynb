```python
import trimesh
import numpy as np
```

```python
#  Loading .obj mesh
mesh_path = "/Users/chaitanyaattanti/Downloads/MIXAR/8samples/explosive.obj"   
mesh = trimesh.load(mesh_path)

#  Extract vertex coordinates
vertices = mesh.vertices 
```

```python
# Stats
print(" TASK - 1 \n")
print(f"Number of vertices: {vertices.shape[0]}")

axis_labels = ["X", "Y", "Z"]

for i in range(3):
    axis_data = vertices[:, i]
    print(f"\n Axis {axis_labels[i]} ")
    print(f"Minimum :  {axis_data.min():4f}")
    print(f"Maximum:  {axis_data.max():.4f}")
    print(f"Mean: {axis_data.mean():.4f}")
    print(f"Std:  {axis_data.std():.4f}")


# Visualize the original mesh

mesh.show()
```
