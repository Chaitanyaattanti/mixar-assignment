```python
import trimesh
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python

mesh_path = "/Users/chaitanyaattanti/Downloads/MIXAR/8samples/cylinder.obj"  

# Load mesh
mesh = trimesh.load(mesh_path, process=False)
vertices = mesh.vertices.copy()
faces = mesh.faces.copy()
```

TASK 2: Normalization & Quantization

```python
def plot_vertices(points, title, filename):
    fig = plt.figure(figsize=(7, 6))
    ax = fig.add_subplot(111, projection='3d')
    ax.scatter(points[:, 0], points[:, 1], points[:, 2], s=1)
    ax.set_title(title)

    # Save the plot 
    fig.savefig(filename, dpi=300)
    plt.close()
```

Min–Max Normalization

```python
#  NORMALIZATION METHOD -1( Min–Max Normalization)

vt_min = vertices.min(axis=0)
vt_max = vertices.max(axis=0)

minmax_norm = (vertices - vt_min) / (vt_max - vt_min)

# Save normalized mesh
minmax_norm_mesh = trimesh.Trimesh(vertices=minmax_norm, faces=faces, process=False)
minmax_norm_mesh.export("Min_Max_norm.ply")


plot_vertices(minmax_norm, "Min – Max Normalized Mesh", "Min_Max_norm.png")
```

Quantization for min–max

```python
# Quantization for min–max normalized mesh

BINS = 1024
minmax_quantized = np.floor(minmax_norm * (BINS - 1)).astype(np.int32)

# Converting  to float for saving because ply can't store int for vertices
minmax_quantized_float = minmax_quantized.astype(float)

minmax_quant_mesh = trimesh.Trimesh(vertices=minmax_quantized_float, faces=faces, process=False)
minmax_quant_mesh.export("Min_Max_quantized.ply")

plot_vertices(minmax_quantized_float, "Min–Max Quantized Mesh", "Min_Max_quantized.png")
```

Unit Sphere Normalization

```python
# center of the mesh
center = vertices.mean(axis=0)
centered = vertices - center

# find the max distance to origin i.e, radius 
rad = np.max(np.linalg.norm(centered, axis=1))

#scaling to unit sphere
unit_norm = centered / rad

# Saving normalized mesh
unit_norm_mesh = trimesh.Trimesh(vertices=unit_norm, faces=faces, process=False)
unit_norm_mesh.export("Unit_sphere_normalized.ply")


plot_vertices(unit_norm, "Unit Sphere Normalized Mesh", "Unit_sphere_normalized.png")
```

Quantization for Unit Sphere

```python

#  Quantization 
# For quantization range should be [0,1], so shifting [-1,1] to  [0,1]
shifted_unit_norm = (unit_norm + 1) / 2

unit_quantized = np.floor(shifted_unit_norm * (BINS - 1)).astype(np.int32)
unit_quantized_float = unit_quantized.astype(float)

# Save quantized mesh
unit_quant_mesh = trimesh.Trimesh(vertices=unit_quantized_float, faces=faces, process=False)
unit_quant_mesh.export("unit_sphere_quantized.ply")


plot_vertices(unit_quantized_float, "Unit Sphere Quantized Mesh", "Unit_sphere_quantized.png")
```
